```python
rows = []
while True:

    
    element = driver.find_element(By.ID, 'DataTables_Table_0')
    原始文字 = element.text
    依橫排分割後的文字 = 原始文字.split('\n')
    欄位名稱 = 依橫排分割後的文字[0].split(' ')

    for 橫排文字 in 依橫排分割後的文字[1:]:
        row = 橫排文字.split(' ')
        row = row[1:]
        if not len(row) == len(欄位名稱):
            row = row[:3] + [''] + row[3:]
        rows = rows + [row]

    
    element = driver.find_element(By.XPATH, "//*[text()='下一頁']")

    
    if element.get_attribute("class") == "aspNetDisabled":
        break

    element.click()
    time.sleep(1)
```

```python
import pandas as pd

df = pd.DataFrame(rows, columns= 欄位名稱)

for col_name in ['成交量','平均價','上價','中價','下價','最高價','最低價',]:
    df[col_name] = df[col_name].astype('float')

df.to_csv("./蔬菜價格_20241101_p2.csv")
print(df)
```
